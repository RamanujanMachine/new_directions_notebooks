## 1. Rivoal: unresolved recurrence

[Source, Theorem 2 and section 6.2](https://rivoal.perso.math.cnrs.fr/articles/gamma.pdf).
The printed recurrence has B(0) = 0, forcing the two solutions to become proportional.
Its constant quotient differs from the target. A corrected recurrence and valid seeds are needed.

```python
n = sp.Symbol('n', integer=True, nonnegative=True)
A = 3*(n+2)*(27*n**2+36*n+13)*(3*n+4)**3
B = 162*n*(2*n+1)*(27*n**2+90*n+76)*(n+1)**2
C = 10935*n**6+80190*n**5+238545*n**4+368550*n**3+312552*n**2+138648*n+25408
p0, p1 = sp.Rational(1, 2), sp.Rational(93, 10)
q0, q1 = sp.Rational(1, 9), sp.Rational(13, 27)
assert B.subs(n, 0) == 0
p2_printed = (C.subs(n, 0)*p1-B.subs(n, 0)*p0)/A.subs(n, 0)
q2_printed = (C.subs(n, 0)*q1-B.subs(n, 0)*q0)/A.subs(n, 0)
assert p2_printed/q2_printed == p1/q1
print('Printed recurrence forces p2/q2 =', p2_printed/q2_printed)
print('Target Gamma(1/3)^3 =', sp.N(sp.gamma(sp.Rational(1, 3))**3, 20))
```
